```python
df["Genre"].value_counts().plot(kind="bar")
plt.title("Movies by Genre")
plt.show()
```

```python
df["Language"].value_counts().plot(kind="pie",autopct="%1.1f%%")
plt.show()
```

```python
plt.hist(df["Rating"],bins=10)
plt.xlabel("Rating")
plt.show()
```

```python
plt.hist(df["Revenue_M"],bins=10)
plt.show()
```

```python
plt.scatter(df["Budget_M"],df["Revenue_M"])
plt.xlabel("Budget")
plt.ylabel("Revenue")
plt.show()
```

```python
sns.boxplot(x="Genre",y="Rating",data=df)
plt.xticks(rotation=45)
plt.show()
```
