# Adult Data

```python
adult = pd.read_csv("adult.csv")
```

```python
adult.head()
```

```python
adult.shape
adult.columns
adult.info()
adult.describe()
adult.describe(include="object")
```

```python
adult.isnull().sum()
```

```python
plt.figure(figsize=(12,6))
sns.heatmap(adult.isnull(), cbar=True)
plt.title("Missing Values")
plt.show()
```

```python
adult.duplicated().sum()
```

```python
adult.drop_duplicates(inplace=True)
```

```python
adult["workclass"].value_counts()
```

```python
sns.countplot(x="workclass", data=adult)
plt.xticks(rotation=45)
plt.show()
```

```python
import matplotlib.pyplot as plt

counts = adult["workclass"].value_counts()

def autopct_format(pct):
    return f"{pct:.1f}%" if pct >= 1 else ""   # Hide percentages below 1%

plt.figure(figsize=(8,8))

plt.pie(
    counts,
    labels=None,
    autopct=autopct_format,
    startangle=190
)

plt.legend(counts.index, loc="upper left", bbox_to_anchor=(1.02, 1))

plt.title("Workclass Distribution")
plt.tight_layout()
plt.show()
```

```python
sns.histplot(adult["age"], kde=True)
plt.show()
```

```python
sns.boxplot(x=adult["age"])
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(adult.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.show()
```

```python
adult = pd.get_dummies(adult, drop_first=True)
```
