```python
store = [{'item': 'Rice', 'price':300, 'quantity': 1000 },
        {'item': 'Flour', 'price':300, 'quantity': 1000 },  
        {'item': 'Oil', 'price':600, 'quantity': 1000 },
        {'item': 'Soap', 'price':300, 'quantity': 1000 },
        {'item': 'Shampoo', 'price':300, 'quantity': 1000 },  
        {'item': 'Sugar', 'price':170, 'quantity': 1000 },
        {'item': 'Surf', 'price':300, 'quantity': 1000 },
        {'item': 'Milk', 'price':240, 'quantity': 1000 },
        {'item': 'Tea', 'price':700, 'quantity': 1000 },
        {'item': 'Egg', 'price':25, 'quantity': 1000 },
        {'item': 'Bread', 'price':300, 'quantity': 1000 },
        ]

cart = []

while True:  
    product = input("\nWhat do you want? (or type 'done' to finish): ")
    
    if product.lower() == 'done':
        break
    
    found = False
    for pro in store:
        if pro['item'].lower() == product.lower():  # Case-insensitive matching
            print(' Available')
            print(f' {pro["item"]} - {pro["price"]} Rupees per unit')
            print(f' Available quantity: {pro["quantity"]}')
            
            quant = int(input('How much do you want? '))
            
            if quant <= pro['quantity']:
                amount = pro['price'] * quant
                pro['quantity'] -= quant
                cart.append({"item": pro['item'], "amount": amount, "quantity": quant})
                print(f' Added {quant} {pro["item"]}(s) to cart. Total: {amount} Rupees')
                found = True
                break
            else:
                print(f' Sorry! Only {pro["quantity"]} {pro["item"]} available, not {quant}')
                found = True
                break
    
    if not found:
        print(' Not Available in our store')

# Calculate totals
total_bill = sum(item['amount'] for item in cart)
total_items = sum(item['quantity'] for item in cart)

# Stylish Invoice
print("\n" + "="*50)
print(" MEGA MART - SHOPPING INVOICE  ".center(50))
print("="*50)
print(f"{'Date:':<15} {__import__('datetime').datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
print(f"{'Bill No:':<15} INV-{__import__('random').randint(1000,9999)}")
print("-"*50)
print(f"{'Item':<20} {'Qty':<8} {'Price':<10} {'Total':<10}")
print("-"*50)

for item in cart:
    unit_price = item['amount'] // item['quantity']
    print(f"{item['item']:<20} {item['quantity']:<8} ${unit_price:<9} ${item['amount']:<9}")

print("-"*50)
print(f"{'Total Items:':<20} {total_items:<8} {'Subtotal:':<10} ${total_bill:<9}")
print(f"{'Tax (5%):':<20} {'':<8} {'':<10} ${total_bill * 0.05:<9.2f}")
print(f"{'Grand Total:':<20} {'':<8} {'':<10} ${total_bill * 1.05:<9.2f}")
print("="*50)
print(" THANK YOU FOR SHOPPING WITH US! ".center(50))
print(" Visit Again! Have a Great Day! ".center(50))
print("="*50)

print(f"\n Your cart summary: {cart}")
```
